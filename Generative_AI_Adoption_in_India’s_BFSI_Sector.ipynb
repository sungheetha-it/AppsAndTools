```python
import matplotlib.pyplot as plt
import seaborn as sns

# Calculate average GenAI Adoption Rate by Sector
adoption_by_sector = df.groupby('Sector')['GenAI_Adoption_Rate'].mean().reset_index()

fig = plt.figure(figsize=(10, 6))
sns.barplot(x='Sector', y='GenAI_Adoption_Rate', data=adoption_by_sector, palette='viridis')
plt.title('Average Generative AI Adoption Rate by BFSI Sector in India')
plt.xlabel('BFSI Sector')
plt.ylabel('Average GenAI Adoption Rate')
plt.ylim(0, adoption_by_sector['GenAI_Adoption_Rate'].max() * 1.2) # Set y-axis limit for better visualization
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```
