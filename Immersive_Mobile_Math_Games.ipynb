## Visualizing a Hypothetical Insight: Impact of Immersive Mobile Math Games

This section generates and visualizes hypothetical data to illustrate a potential insight from the research proposal: that immersive mobile math games can improve functional independence in children with intellectual disabilities. We will simulate 'Functional Independence Scores' for a control group and a game-based intervention group, both before and after an intervention period.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set a random seed for reproducibility
np.random.seed(42)

# Number of participants per group
n_participants = 50

# Hypothetical 'Functional Independence Score' (e.g., on a scale of 0-100)
# Control Group: Little to no improvement expected
control_pre = np.random.normal(loc=40, scale=10, size=n_participants)
control_post = control_pre + np.random.normal(loc=2, scale=5, size=n_participants) # Small, noisy improvement

# Game-Based Group: Significant improvement expected
game_pre = np.random.normal(loc=40, scale=10, size=n_participants)
game_post = game_pre + np.random.normal(loc=15, scale=8, size=n_participants) # Larger improvement

# Ensure scores stay within a plausible range (e.g., 0-100)
control_pre = np.clip(control_pre, 0, 100)
control_post = np.clip(control_post, 0, 100)
game_pre = np.clip(game_pre, 0, 100)
game_post = np.clip(game_post, 0, 100)

# Create DataFrame
data = pd.DataFrame({
    'Group': ['Control'] * n_participants * 2 + ['Game-Based'] * n_participants * 2,
    'Time': ['Pre-Intervention'] * n_participants + ['Post-Intervention'] * n_participants +
            ['Pre-Intervention'] * n_participants + ['Post-Intervention'] * n_participants,
    'Score': np.concatenate([control_pre, control_post, game_pre, game_post])
})

display(data.head())
print(data.info())
```

```python
plt.figure(figsize=(10, 6))
sns.barplot(data=data, x='Time', y='Score', hue='Group', palette={'Control': 'lightcoral', 'Game-Based': 'mediumseagreen'})

plt.title('Hypothetical Functional Independence Scores: Control vs. Game-Based Intervention')
plt.xlabel('Intervention Period')
plt.ylabel('Functional Independence Score')
plt.ylim(0, 100) # Set Y-axis limits to match score range
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Intervention Group')
plt.show()
```

```python
plt.figure(figsize=(12, 7))
sns.boxplot(data=data, x='Time', y='Score', hue='Group', palette={'Control': 'lightcoral', 'Game-Based': 'mediumseagreen'})

plt.title('Distribution of Functional Independence Scores by Group and Time (Boxplot)')
plt.xlabel('Intervention Period')
plt.ylabel('Functional Independence Score')
plt.ylim(0, 100) # Maintain consistent y-axis limits
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Intervention Group')
plt.show()
```

### Interpretation of the Boxplot

The boxplot provides a more detailed view of the score distribution:

*   **Boxes**: Represent the interquartile range (IQR), containing the middle 50% of the data.
*   **Lines inside boxes**: Indicate the median score.
*   **Whiskers**: Extend to 1.5 times the IQR from the upper and lower quartiles, showing the spread of the majority of the data.
*   **Points outside whiskers**: Represent outliers.

From this boxplot, you can observe:

*   **Control Group**: The median and overall distribution of scores remain relatively consistent from pre- to post-intervention, with possibly a slight upward shift in the median.
*   **Game-Based Group**: A clear upward shift in the median score and the entire distribution from pre- to post-intervention, indicating a significant increase in functional independence scores. The spread (IQR) might also change, potentially becoming narrower or wider depending on the intervention's consistency.

This plot reinforces the hypothetical insight that the immersive mobile math games could lead to a substantial improvement in functional independence, with a better understanding of the data's variance and skewness.
