## Neural Implicit Representations

Neural Implicit Representations (NIRs) use neural networks to represent complex functions, such as 3D shapes, images, or scenes. Instead of storing explicit data structures (like meshes or pixel grids), NIRs learn a continuous mapping from coordinates (e.g., `(x, y, z)` for a 3D point) to a property (e.g., occupancy, color, signed distance function value).

This approach offers several benefits:
- **Resolution independence**: The representation is continuous, so it can be queried at any resolution.
- **Compactness**: A small neural network can represent highly detailed geometries or functions.
- **Differentiability**: The neural network is differentiable, which allows for optimization techniques to reconstruct or manipulate the represented data.

Below is a basic Python code template using `PyTorch` to define a simple Multi-Layer Perceptron (MLP) as an implicit representation. This example can be adapted for various applications like representing a 2D image or a 3D shape.

```
# First, ensure PyTorch is installed
# !pip install torch

import torch
import torch.nn as nn
import torch.nn.functional as F

# Define the Neural Implicit Representation (NIR) model
class NeuralImplicit(nn.Module):
    def __init__(self, input_dim=2, output_dim=1, hidden_dim=256, num_layers=4):
        super(NeuralImplicit, self).__init__()

        self.layers = nn.ModuleList()
        self.layers.append(nn.Linear(input_dim, hidden_dim))

        for _ in range(num_layers - 1):
            self.layers.append(nn.Linear(hidden_dim, hidden_dim))

        self.output_layer = nn.Linear(hidden_dim, output_dim)

        # Changed activation function from ReLU to GELU for potentially smoother outputs
        self.activation = nn.GELU()

    def forward(self, x):
        # x is the input coordinate (e.g., (x,y) for 2D, (x,y,z) for 3D)
        for layer in self.layers:
            x = self.activation(layer(x))
        x = self.output_layer(x)
        return x

# --- Example Usage ---

# 1. Initialize the model
# For a 2D image (input_dim=2 for x,y coordinates, output_dim=3 for R,G,B color)
# Increased num_layers slightly for demonstration
image_nir = NeuralImplicit(input_dim=2, output_dim=3, hidden_dim=128, num_layers=6)
print("2D Image NIR Model:\n", image_nir)

# For a 3D shape (input_dim=3 for x,y,z coordinates, output_dim=1 for occupancy or SDF)
# Increased num_layers slightly for demonstration
shape_nir = NeuralImplicit(input_dim=3, output_dim=1, hidden_dim=256, num_layers=10)
print("\n3D Shape NIR Model:\n", shape_nir)

# 2. Generate some example input coordinates
# For a 2D image, let's create a batch of 100 random (x,y) coordinates
coordinates_2d = torch.rand(100, 2) * 2 - 1 # Coordinates between -1 and 1

# For a 3D shape, a batch of 50 random (x,y,z) coordinates
coordinates_3d = torch.rand(50, 3) * 2 - 1 # Coordinates between -1 and 1

# 3. Pass coordinates through the model to get the implicit output
output_2d = image_nir(coordinates_2d)
output_3d = shape_nir(coordinates_3d)

print("\nOutput for 2D Image NIR (first 5 values):\n", output_2d[:5])
print("\nOutput for 3D Shape NIR (first 5 values):\n", output_3d[:5])

# Further steps would involve:
# - Defining a loss function (e.g., MSE for image, BCE for occupancy, L1/L2 for SDF)
# - Creating an optimizer (e.g., Adam)
# - Training loop to optimize the network weights to represent the target data
# - Sampling dense grids of coordinates to visualize the learned implicit function
```

### 1. Generate a Sample 2D Dataset (e.g., a Sine Wave)

```
import numpy as np
import matplotlib.pyplot as plt

# Function to generate a 2D sine wave dataset
def generate_2d_sine_wave(num_samples=1000):
    # Generate random x, y coordinates between -1 and 1
    x = torch.rand(num_samples, 1) * 2 - 1
    y = torch.rand(num_samples, 1) * 2 - 1

    # Combine into a single coordinate tensor
    coords = torch.cat([x, y], dim=1)

    # Calculate the target value using a 2D sine wave function
    target = torch.sin(coords[:, 0]) + torch.cos(coords[:, 1])

    # Reshape target to be (num_samples, 1)
    target = target.unsqueeze(1)

    return coords, target

# Generate the dataset
train_coords, train_targets = generate_2d_sine_wave(num_samples=5000)

print(f"Training coordinates shape: {train_coords.shape}")
print(f"Training targets shape: {train_targets.shape}")

# Visualize a small subset of the generated data
fig = plt.figure(figsize=(8, 6))
ax = fig.add_subplot(111, projection='3d')
ax.scatter(train_coords[:, 0].numpy(), train_coords[:, 1].numpy(), train_targets[:, 0].numpy(), cmap='viridis', s=5)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Value')
ax.set_title('Sample 2D Sine Wave Dataset')
plt.show()
```

### 2. Define Loss Function, Optimizer, and Training Loop

```
from torch.optim import Adam

# Re-initialize the model for this specific task (2D input, 1D output)
# Increased num_layers to 8 for better capacity
nir_model = NeuralImplicit(input_dim=2, output_dim=1, hidden_dim=256, num_layers=8)

# Define loss function (Mean Squared Error for regression)
criterion = nn.MSELoss()

# Define optimizer
optimizer = Adam(nir_model.parameters(), lr=0.001)

# Training parameters
num_epochs = 2000

print("Starting training...")
for epoch in range(num_epochs):
    # Forward pass
    outputs = nir_model(train_coords)
    loss = criterion(outputs, train_targets)

    # Backward and optimize
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    if (epoch + 1) % 200 == 0:
        print(f'Epoch [{epoch+1}/{num_epochs}], Loss: {loss.item():.4f}')

print("Training finished.")
```

### 3. Visualize the Trained Model's Output

```
# Generate a grid of points to evaluate the trained model
resolution = 100
x_linspace = torch.linspace(-1, 1, resolution)
y_linspace = torch.linspace(-1, 1, resolution)
x_grid, y_grid = torch.meshgrid(x_linspace, y_linspace, indexing='ij')

# Flatten the grids and combine into coordinates
eval_coords = torch.stack([x_grid.flatten(), y_grid.flatten()], dim=1)

# Pass through the trained model
with torch.no_grad():
    predicted_values = nir_model(eval_coords).reshape(resolution, resolution).numpy()

# Visualize the predicted surface
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

surface = ax.plot_surface(x_grid.numpy(), y_grid.numpy(), predicted_values,
                          cmap='viridis', edgecolor='none')
fig.colorbar(surface, shrink=0.5, aspect=5)
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_zlabel('Predicted Value')
ax.set_title('Trained Neural Implicit Representation of 2D Sine Wave')
plt.show()
```
