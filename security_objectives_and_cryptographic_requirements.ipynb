```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Define security objectives and cryptographic requirements
security_objectives = [
    'Confidentiality',
    'Integrity',
    'Authenticity',
    'Non-repudiation'
]
cryptographic_requirements = [
    'Encryption',
    'Hashing',
    'Digital Signatures',
    'Key Exchange'
]

# Create a relationship matrix (1 if related, 0 if not)
# This is a simplified representation.
# For example, Encryption primarily supports Confidentiality.
# Hashing supports Integrity.
# Digital Signatures support Integrity, Authenticity, and Non-repudiation.
# Key Exchange supports Authenticity (e.g., mutual authentication) and enables Confidentiality.
data = [
    [1, 0, 0, 0],  # Confidentiality: Encryption
    [0, 1, 1, 0],  # Integrity: Hashing, Digital Signatures
    [0, 0, 1, 1],  # Authenticity: Digital Signatures, Key Exchange
    [0, 0, 1, 0]   # Non-repudiation: Digital Signatures
]

# Create a DataFrame for better labeling and plotting
df_relationships = pd.DataFrame(data, index=security_objectives, columns=cryptographic_requirements)

# Plotting the heatmap
plt.figure(figsize=(10, 6))
sns.heatmap(
    df_relationships,
    annot=True,      # Show the data values on the heatmap
    cmap='viridis',  # Color map
    fmt='d',         # Format annotation as integers
    linewidths=.5,   # Add lines between cells
    linecolor='black'
)

plt.title('Relationship between Security Objectives and Cryptographic Requirements')
plt.xlabel('Cryptographic Requirements')
plt.ylabel('Security Objectives')
plt.tight_layout()
plt.show()
```
